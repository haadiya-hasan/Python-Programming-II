### **Create the tensors**

```python
# Tensors is the core data structure in pytorch, it is quite similar to NumPy arrays

x = torch.tensor([1, 2, 3])
print(x)
```

```python
# To confirm if this is a tensor or not
type(x)
```

```python
# Using the "empty" command. It creates a tensor of a specified shape filled with uninitialized data.

a = torch.empty(2, 3)
print(a)
```

```python
# To make a tensor purely out of 0s we use "zeros".

torch.zeros(2,3)
```

```python
# To make a tensor purely out of 1s we use "ones".

torch.ones(2,3)
```

```python
# "Rand" in PyTorch is used to create a tensor of a given shape filled with random numbers from a uniform distribution on the interval.

torch.rand(6, 2)
```

```python
# "Manual_seed" used to set the seed for the random number generator (RNG) across all devices, including the CPU and CUDA.
# We use it with torch.rand() to creates random number (it basically provides a structure for the seeds to be generated).

torch.manual_seed(500)
torch.rand(2, 3)
```

```python
# One more example

torch.manual_seed(500)
torch.rand(4,5)
```

```python
# Other Useful Tensor Creation Commands in PyTorch

# torch.arange(start, end, step) -> Creates a tensor with values starting from 'start' up to 'end-1' using the given step size.
print("\n using arange ->", torch.arange(0,10,2))

# torch.linspace(start, end, steps) -> Creates a tensor with a fixed number of evenly spaced values between start and end.
print("\n using linspace ->", torch.linspace(0, 10, 7))

# torch.eye(n) -> Creates an n × n identity matrix with 1s on the diagonal and 0s elsewhere.
print("\n using eye ->", torch.eye(5))

# torch.full(size, value) -> Creates a tensor of the given shape filled entirely with the specified value.
print("\n using full ->", torch.full((3, 3), 5))
```

# **Tensor Shape and Tensor Creation in PyTorch**

### Creating Tensors in Different Ways

```python
torch.tensor([[1,2,3],[4,5,6]])

#Creates a tensor from a Python list.
#Shape → (2, 3) → 2 rows and 3 columns.
```

```python
torch.arange(0, 10, 2)

#Creates values from 0 to 10 (excluding 10) with a step size of 2.
# Output → 0, 2, 4, 6, 8
# Useful for generating evenly spaced integer values.
```

```python
torch.linspace(0, 10, 10)

# Creates 10 equally spaced values between 0 and 10 (including both).
# Commonly used in mathematics, plotting, and ML experiments.
```

```python
torch.eye(5)

# Creates a 5 × 5 identity matrix.
# Diagonal elements are 1, all other elements are 0.
# Useful in linear algebra operations.
```

```python
torch.full((3, 3), 5)

# Creates a 3 × 3 tensor filled entirely with 5
# Useful when constant values are needed.
```

```python
# Tensor Shape
x = torch.tensor([[1,2,3],[4,5,6]])
x.shape
```

```python
torch.Size([2, 3])

# Represents the dimensions of the tensor. Here:
# 2 → Number of rows
# 3 → Number of columns
```

```python
torch.empty_like(x)

# Creates a tensor with the same shape as x.
# Values are uninitialized (random garbage values).
# Very fast, but values are meaningless until assigned.
```

```python
torch.zeros_like(x)

# Creates a tensor with the same shape as x, filled with 0s.
```

```python
torch.ones_like(x)

# Creates a tensor with the same shape as x, filled with 1s.
```

```python
torch.rand_like(x, dtype=torch.float32)

# Creates a tensor with the same shape as x.
# Filled with random values between 0 and 1.
# dtype=torch.float32 sets the data type to 32-bit floating point.
```

# **Datatypes and it's Descriptions**

| **Data Type**             | **Dtype**         | **Description**                                                                                                                                                                |
|---------------------------|-------------------|--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **32-bit Floating Point** | `torch.float32`   | Standard floating-point type used for most deep learning tasks. Provides a balance between precision and memory usage.                                                         |
| **64-bit Floating Point** | `torch.float64`   | Double-precision floating point. Useful for high-precision numerical tasks but uses more memory.                                                                               |
| **16-bit Floating Point** | `torch.float16`   | Half-precision floating point. Commonly used in mixed-precision training to reduce memory and computational overhead on modern GPUs.                                            |
| **BFloat16**              | `torch.bfloat16`  | Brain floating-point format with reduced precision compared to `float16`. Used in mixed-precision training, especially on TPUs.                                                |
| **8-bit Floating Point**  | `torch.float8`    | Ultra-low-precision floating point. Used for experimental applications and extreme memory-constrained environments (less common).                                               |
| **8-bit Integer**         | `torch.int8`      | 8-bit signed integer. Used for quantized models to save memory and computation in inference.                                                                                   |
| **16-bit Integer**        | `torch.int16`     | 16-bit signed integer. Useful for special numerical tasks requiring intermediate precision.                                                                                    |
| **32-bit Integer**        | `torch.int32`     | Standard signed integer type. Commonly used for indexing and general-purpose numerical tasks.                                                                                  |
| **64-bit Integer**        | `torch.int64`     | Long integer type. Often used for large indexing arrays or for tasks involving large numbers.                                                                                  |
| **8-bit Unsigned Integer**| `torch.uint8`     | 8-bit unsigned integer. Commonly used for image data (e.g., pixel values between 0 and 255).                                                                                    |
| **Boolean**               | `torch.bool`      | Boolean type, stores `True` or `False` values. Often used for masks in logical operations.                                                                                      |
| **Complex 64**            | `torch.complex64` | Complex number type with 32-bit real and 32-bit imaginary parts. Used for scientific and signal processing tasks.                                                               |
| **Complex 128**           | `torch.complex128`| Complex number type with 64-bit real and 64-bit imaginary parts. Offers higher precision but uses more memory.                                                                 |
| **Quantized Integer**     | `torch.qint8`     | Quantized signed 8-bit integer. Used in quantized models for efficient inference.                                                                                              |
| **Quantized Unsigned Integer** | `torch.quint8` | Quantized unsigned 8-bit integer. Often used for quantized tensors in image-related tasks.                                                                                     |


# **Mathematical Operations**

### 1. Scalar operation

```python
x = torch.rand(2,2)
x
```

```python
# addition
x + 2
# substraction
x - 2
# multiplication
x * 3
# division
x / 3
# int division
(x * 100)//3
# mod
((x * 100)//3)%2
# power
x**2
```

### 2. Element wise operation

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
# add
a + b
# sub
a - b
# multiply
a * b
# division
a / b
# power
a ** b
# mod
a % b
```

```python
c = torch.tensor([1, -2, 3, -4])
```

```python
# abs
torch.abs(c)
```

```python
# negative
torch.neg(c)
```

```python
d = torch.tensor([1.9, 2.3, 3.7, 4.4, 1.5])
```

```python
# round
torch.round(d)
```

```python
# ceil
torch.ceil(d)
```

```python
# floor
torch.floor(d)
```

### 3. Reduction operation

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
```

```python
torch.sum(e, dim=0)
```

```python
# sum
torch.sum(e)

# adds all the elements of the tensor together and returns the total sum.
```

```python
# mean
torch.mean(e)
# mean along col

torch.mean(e, dim=0)
```

```python
# median
torch.median(e)
```

```python
# max and min
torch.max(e)
torch.min(e)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax
torch.argmax(e)
```

```python
# argmin
torch.argmin(e)
```

### 4. Matrix operations

```python
f = torch.randint(size=(2,3), low=0, high=10)
g = torch.randint(size=(3,2), low=0, high=10)

print(f)
print(g)
```

```python
# matrix multiplcation
torch.matmul(f, g)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
torch.dot(vector1, vector2)
```

```python
# transpose
torch.transpose(f, 0, 1)
```

```python
h = torch.randint(size=(3,3), low=0, high=10, dtype=torch.float32)
print(h)
```

```python
# determinant
torch.det(h)
```

```python
# inverse
torch.inverse(h)
```

### 5. Comparison operations

```python
i = torch.randint(size=(2,3), low=0, high=10)
j = torch.randint(size=(2,3), low=0, high=10)

print(i)
print(j)
```

```python
# greater than
i > j
# less than
i < j
# equal to
i == j
# not equal to
i != j
# greater than equal to

# less than equal to
```

### 6. Special functions

```python
k = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
print(k)
```

```python
# log
torch.log(k)
```

```python
# exp
torch.exp(k)
```

```python
# sqrt
torch.sqrt(k)
```

```python
# sigmoid scaling bw 0 and 1
torch.sigmoid(k)
```

```python
# softmax scaling bw 0 and 1 with sum of all elements as 1
torch.softmax(k, dim=0)
```
